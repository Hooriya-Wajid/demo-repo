Slicing a List​

```python
players=['charles', 'martina', 'michael', 'florence', 'eli']
print(players[0:3])
print(players[0:4])
print(players[1:4])
print(players[:4])
print(players[2:])
print(players[-3:])
```

```python
players=['charles', 'martina', 'michael', 'florence', 'eli']
print("Here are the first three players on my team")
for player in players[:3]:
  print(player.title())
```

Copying a List​

```python
my_foods=['pizza','falafel','carrotcake']
print(my_foods)
friend_foods=my_foods[:]
print(friend_foods)
print("My friend's favourite foods are:")
print(friend_foods)
```

```python
my_foods=['pizza','falafel','carrotcake']
print("My favourite foods are:")
print(my_foods)
friend_foods=my_foods[:]

print("My friend's favourite foods are:")
print(friend_foods)
```

```python
my_foods=['pizza','falafel','carrotcake']
friend_foods=my_foods[:]
my_foods.append('cannoli')
friend_foods.append('ice cream')
print("My favourite foods are")
print(my_foods)
print("My friend's favourite foods are")
print(friend_foods)
```

This does not create a new list.

 Instead, it makes friend_foods refer to the same list as my_foods.

Both variables point to the same object in memory.

```python
my_foods=['pizza','falafel','carrotcake']
friend_foods=my_foods
my_foods.append('cannoli')
friend_foods.append('ice cream')
print("My favourite foods are:")
print(my_foods)
print("\nMy friend's favourite foods are:")
print(friend_foods)
```

Easy way to remember

friend_foods = my_foods → One list, two names (aliases).

friend_foods = my_foods[:] → Two independent lists (a copy).

This is one of the most important concepts when learning Python lists.

Looping Through All Values in a Tuple​

```python
dimensions=(200,50)
for dimension in dimensions:
  print(dimension)
```

Writing Over a Tuple​

```python
dimensions=(200,50)
print("Orignal dimensions:")
for dimension in dimensions:
  print(dimension)

dimensions=(400,100)
print("\nModified dimensions:")
for dimension in dimensions:
  print(dimensions)
```

Converting a list to a tuple and vice versa​

```python
number_list=[1,2,3,4,5]
tup=tuple(number_list)
print(tup)
#converting tuple again into list
lst=list(tup)
print(lst)
```

Converting a list to a tuple and vice versa​

```python
#indexing
#if you mistakenly write [] instead of () this will produce an error here then
numbers=(1,2,3,4)
print(numbers[0])
print(numbers[1])
print(numbers)

#slicing
subset=numbers[1:3]
print(subset)

#length
len(numbers)

#Concatenation: Combine tuples.​
new_tuple= numbers+(5,6)
print(new_tuple)

#Unpacking: Assign tuple elements to variables.
a,b,c,d=numbers
print(numbers)
print(a)
print(b)
print(c)
print(d)
```

Unpacking means taking the values from a tuple and storing each value in a separate variable.

Example:

numbers = (10, 20, 30, 40)

a, b, c, d = numbers

After unpacking:

a → 10
b → 20
c → 30
d → 40
