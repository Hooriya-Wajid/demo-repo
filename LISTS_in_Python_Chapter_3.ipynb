Accessing Elements in the list.

Lists are ordered collections, so any element in a list by telling Python the
position, or index, of the item desired​



```python
bicycles=['trek', 'cannondale', 'redline', 'specialized']
print(bicycles[0])
print(bicycles[0].title())
print(bicycles[1])
print(bicycles[2])
print(bicycles[3])
print(bicycles[-1])
print(bicycles[-2])
print(bicycles[-3])
print(bicycles[-4])
```

Removing an Item Using the pop() Method​

```python
motorcycles=['Honda', 'Yamaha', 'Suzuki']
print(motorcycles)
popped_motorcycle=motorcycles.pop()
print(motorcycles)
print(popped_motorcycle)
```

'suzuki' was removed from the end of the list and is now assigned to the variable
popped_motorcycle​

```python
motorcycles=['Honda', 'Yamaha','Suzuki']
last_owned=motorcycles.pop()
#reason for the syntax must be explained
print(f"The last motorcycle I owned was a {last_owned.title()}.")
```

If you only know the value of the item you want to remove,

you can use the remove()
method​

For example, say we want to remove the value 'ducati' from the list of motorcycles

```python
motorcycles=['Honda', 'Yamaha', 'Suzuki','ducati']
print(motorcycles)
motorcycles.remove('ducati')
print(motorcycles)
```

Removing an Item by Value​

You can also use the remove() method to work with a value that’s being removed
from a list​

```python
motorcycles=['Honda','Yamaha','Suzuki','Ducati']
print(motorcycles)
too_expensive='Ducati'
motorcycles.remove(too_expensive)
print(motorcycles)
print(f"\nA {too_expensive.title()} is too expesnive for me")
```

Here's the logic of each line very briefly:

motorcycles = ['Honda', 'Yamaha', 'Suzuki', 'Ducati']

Logic: Creates a list of four motorcycle brands.

print(motorcycles)

Logic: Displays the complete list.

too_expensive = 'Ducati'

Logic: Stores the item to be removed in a variable.

motorcycles.remove(too_expensive)

Logic: Removes the value stored in too_expensive ("Ducati") from the list.

print(motorcycles)

Logic: Displays the updated list after removal.

print(f"\nA {too_expensive.title()} is too expesnive for me")

Logic: Prints a message using the removed value.

f"..." allows inserting variables into the string.

\n starts the output on a new line.

.title() capitalizes the first letter of the word.
Output:
['Honda', 'Yamaha', 'Suzuki', 'Ducati']
['Honda', 'Yamaha', 'Suzuki']

A Ducati is too expensive for me

Key concept: The variable too_expensive still stores "Ducati" even after it has been removed from the list, so you can still use it later in the program.

Organizing a list

Sorting a List Permanently with the sort() Method​

Imagine we have a list of cars and want to change the order of the list to store them alphabetically​

```python
cars=["bmw","audi", "toyota", "subaru"]
cars.sort()
print(cars)
```

You can also sort this list in reverse-alphabetical order

by passing the​ argument reverse=True to the sort() method​



```python
cars=["bmw", "audi", "toyota","subaru"]
cars.sort(reverse=True)
print(cars)
```

Sorting a List Temporarily with the sorted() Function​

```python
cars=["bmw", "audi", "toyota","subaru"]
print("Here's the orignal list:")
print(cars)
print("Here's the sorted list:")
print(sorted(cars))
print("\nHere's the orignal list again:")
print(cars)
cars.reverse()
print(cars)
```
